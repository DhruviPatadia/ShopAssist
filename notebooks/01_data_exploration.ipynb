```python
import pandas as pd

documents = []

for file_path in sorted(policy_dir.glob("*.md")):
    text = file_path.read_text(encoding="utf-8")

    documents.append({
        "source": file_path.name,
        "text": text
    })

documents_df = pd.DataFrame(documents)

print(f"Loaded {len(documents_df)} documents.")
documents_df.head()
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

```python
document_embeddings = embedding_model.encode(
    documents_df["text"].tolist(),
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embedding shape:", document_embeddings.shape)
```

```python
import faiss

embedding_dimension = document_embeddings.shape[1]

index = faiss.IndexFlatIP(embedding_dimension)

index.add(document_embeddings)

print("FAISS index created.")
print("Number of indexed documents:", index.ntotal)
```

```python
query = "What's the deadline for sending something back?"

query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
)

scores, indices = index.search(query_embedding, k=3)

for rank, (score, idx) in enumerate(zip(scores[0], indices[0]), start=1):
    print(f"{rank}. {documents_df.iloc[idx]['source']} | similarity: {score:.4f}")
```

```python
def search_policies(query, top_k=3):
    """
    Search the policy knowledge base using semantic similarity.

    Parameters:
        query (str): Customer question
        top_k (int): Number of results to return

    Returns:
        DataFrame containing the most relevant policies and similarity scores.
    """

    # Convert the query into an embedding
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Search FAISS
    scores, indices = index.search(query_embedding, top_k)

    # Build results
    results = []

    for rank, (score, idx) in enumerate(
        zip(scores[0], indices[0]),
        start=1
    ):
        results.append({
            "rank": rank,
            "source": documents_df.iloc[idx]["source"],
            "similarity": round(float(score), 4)
        })

    return pd.DataFrame(results)
```

```python
evaluation_df = pd.read_csv(evaluation_file)

evaluation_results = []

for _, row in evaluation_df.iterrows():

    query = row["question"]
    expected_source = row["expected_source"]

    results = search_policies(query, top_k=3)

    retrieved_sources = results["source"].tolist()

    rank_1 = retrieved_sources[0]
    rank_3 = expected_source in retrieved_sources

    if expected_source is None or pd.isna(expected_source):
        correct_at_1 = None
        correct_at_3 = None
    else:
        correct_at_1 = rank_1 == expected_source
        correct_at_3 = rank_3

    evaluation_results.append({
        "question": query,
        "expected_source": expected_source,
        "rank_1_source": rank_1,
        "rank_1_correct": correct_at_1,
        "correct_at_3": correct_at_3
    })

retrieval_results_df = pd.DataFrame(evaluation_results)

retrieval_results_df
```

```python
print("Total questions:", len(retrieval_results_df))

print("\nRank@1:")
print(
    retrieval_results_df["rank_1_correct"]
    .dropna()
    .mean()
)

print("\nRecall@3:")
print(
    retrieval_results_df["correct_at_3"]
    .dropna()
    .mean()
)
```

```python
reciprocal_ranks = []

for _, row in retrieval_results_df.iterrows():

    expected_source = row["expected_source"]

    if pd.isna(expected_source):
        continue

    results = search_policies(row["question"], top_k=3)
    sources = results["source"].tolist()

    if expected_source in sources:
        rank = sources.index(expected_source) + 1
        reciprocal_ranks.append(1 / rank)
    else:
        reciprocal_ranks.append(0)

mrr = sum(reciprocal_ranks) / len(reciprocal_ranks)

print("Mean Reciprocal Rank (MRR):", round(mrr, 4))
```

```python
def generate_answer(query, top_k=3):
    """
    Generate a customer-support answer using retrieved policy documents.
    Gemini is instructed to answer only from the provided policy context.
    """

    # Retrieve relevant policies
    retrieved = search_policies(query, top_k=top_k)

    # Build policy context
    context_parts = []

    for _, row in retrieved.iterrows():
        source = row["source"]

        # Get the full policy text
        policy_text = documents_df.loc[
            documents_df["source"] == source, "text"
        ].iloc[0]

        context_parts.append(
            f"Source: {source}\n{policy_text}"
        )

    context = "\n\n---\n\n".join(context_parts)

    # Prompt Gemini
    prompt = f"""
You are ShopAssist AI, an e-commerce customer support assistant.

Answer the customer's question using ONLY the policy information
provided below.

Rules:
1. Do not invent information.
2. Do not make assumptions beyond the policies.
3. If the policies do not contain enough information to answer,
   clearly say that the information is not available.
4. Give a concise and helpful customer-support response.
5. Do not mention internal retrieval, embeddings, FAISS, or the
   knowledge base.
6. When appropriate, mention the relevant policy source at the end.

POLICY INFORMATION:
{context}

CUSTOMER QUESTION:
{query}

ANSWER:
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return {
        "answer": response.text,
        "retrieved_documents": retrieved
    }
```

```python
import pandas as pd
import re

def normalize_text(text):
    """Normalize text for simple keyword/phrase matching."""
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = re.sub(r"[^a-z0-9\s\-]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Define key information expected in each answer
expected_keywords = [
    ["30 days"],
    ["30 days"],
    ["damaged"],
    ["contact support"],
    ["3–5 working days", "3-5 working days"],
    ["3–5 working days", "3-5 working days"],
    ["cannot normally be cancelled", "cannot be cancelled"],
    ["cannot normally be cancelled", "cannot be cancelled"],
    ["international shipping"],
    ["international shipping"],
    ["replacement"],
    ["replacement"],
    ["5–10 working days", "5-10 working days"],
    ["debit", "credit"],
    ["one promotional discount code", "one promotional"],
    ["missing", "contact support"],
    ["accidental damage"],
    ["cannot normally be exchanged for cash", "not be exchanged for cash"],
    ["password", "payment card"],
    []  # Out-of-scope question
]


evaluation_checks = []

for i, row in rag_evaluation_df.iterrows():

    generated = normalize_text(row["generated_answer"])

    keywords = expected_keywords[i]

    if not keywords:
        # Out-of-scope question:
        # check whether the system indicates that the information
        # is unavailable rather than inventing an answer.
        out_of_scope_phrases = [
            "do not contain information",
            "does not contain information",
            "information is not available",
            "not available"
        ]

        passed = any(
            phrase in generated
            for phrase in out_of_scope_phrases
        )

    else:
        passed = any(
            normalize_text(keyword) in generated
            for keyword in keywords
        )

    evaluation_checks.append(passed)


rag_evaluation_df["generation_check"] = evaluation_checks

display(
    rag_evaluation_df[
        [
            "question",
            "expected_answer",
            "generated_answer",
            "generation_check"
        ]
    ]
)

print("\nGeneration checks passed:",
      rag_evaluation_df["generation_check"].sum(),
      "/",
      len(rag_evaluation_df))

print(
    "Generation check rate:",
    round(
        rag_evaluation_df["generation_check"].mean() * 100,
        2
    ),
    "%"
)
```

```python
conversation_history = []

print("Conversation memory initialized.")
```

```python
def add_to_memory(user_message, assistant_message):
    """
    Store one user-assistant interaction in conversation memory.
    """

    conversation_history.append({
        "role": "user",
        "content": user_message
    })

    conversation_history.append({
        "role": "assistant",
        "content": assistant_message
    })


def show_memory():
    """
    Display the current conversation history.
    """

    for message in conversation_history:
        role = message["role"].capitalize()
        print(f"{role}: {message['content']}\n")
```

```python
add_to_memory(
    "Can I return an item?",
    "Yes. Eligible products can normally be returned within 30 days of delivery."
)

add_to_memory(
    "What about a damaged item?",
    "If an item arrives damaged, please contact support and provide photographs if requested."
)

show_memory()
```

```python
def generate_answer_with_memory(query, top_k=3):
    """
    Prepare a grounded RAG prompt that includes conversation history.
    Does not call Gemini until explicitly invoked.
    """

    # Retrieve relevant policies for the current question
    retrieved = search_policies(query, top_k=top_k)

    # Build policy context
    context_parts = []

    for _, row in retrieved.iterrows():
        source = row["source"]

        policy_text = documents_df.loc[
            documents_df["source"] == source, "text"
        ].iloc[0]

        context_parts.append(
            f"Source: {source}\n{policy_text}"
        )

    context = "\n\n---\n\n".join(context_parts)

    # Include previous exchanges
    history_text = "\n".join(
        f"{message['role'].capitalize()}: {message['content']}"
        for message in conversation_history
    )

    prompt = f"""
You are ShopAssist AI, an e-commerce customer support assistant.

Use the conversation history to understand follow-up questions.
Use ONLY the policy information below to answer policy questions.
Do not invent information. If the policies do not answer the
question, clearly say that the information is unavailable.

CONVERSATION HISTORY:
{history_text if history_text else "No previous conversation."}

RELEVANT POLICIES:
{context}

CURRENT CUSTOMER QUESTION:
{query}

Provide a concise, helpful answer and cite the relevant policy source.
"""

    # Return the prepared prompt without making an API request
    return {
        "prompt": prompt,
        "retrieved_documents": retrieved
    }


print("Memory-aware RAG function defined successfully.")
```

```python
def search_policies_with_memory(query, top_k=3):
    """
    Search primarily using the current question.
    Use recent history only as a fallback if needed.
    """

    # First, search using the current question alone
    results = search_policies(query, top_k=top_k)

    # If the current question already finds a relevant policy,
    # keep that ranking rather than mixing in all previous topics.
    if not results.empty and results.iloc[0]["similarity"] >= 0.60:
        return results

    # Otherwise, try adding recent conversation context
    recent_messages = conversation_history[-4:]

    history_context = " ".join(
        message["content"]
        for message in recent_messages
    )

    contextual_query = (
        f"Recent context: {history_context}. "
        f"Current question: {query}"
    )

    return search_policies(contextual_query, top_k=top_k)


print("Updated context-aware retrieval ready.")
```

```python
follow_up = "What if it arrived broken?"

results = search_policies_with_memory(follow_up)

print("Follow-up question:", follow_up)
display(results)
```

```python
def resolve_follow_up_query(query):
    """
    Resolve common follow-up questions using recent conversation.
    Uses simple rules only; no API calls.
    """

    query_lower = query.lower()

    # Handle follow-ups about broken or damaged items
    damage_terms = [
        "broken", "damaged", "cracked", "defective"
    ]

    if any(term in query_lower for term in damage_terms):
        return f"damaged item arrived broken: {query}"

    # Handle follow-ups about returns
    if any(term in query_lower for term in ["return", "send back"]):
        return f"product return policy: {query}"

    # Handle follow-ups about refunds
    if any(term in query_lower for term in ["refund", "money back"]):
        return f"refund policy: {query}"

    # Default: keep the original query
    return query


def search_policies_with_memory(query, top_k=3):
    resolved_query = resolve_follow_up_query(query)
    return search_policies(resolved_query, top_k=top_k)


print("Follow-up query resolver ready.")
```

```python
follow_up = "What if it arrived broken?"

print("Original question:", follow_up)
print("Resolved question:", resolve_follow_up_query(follow_up))

display(search_policies_with_memory(follow_up))
```

```python
def prepare_answer_with_memory(query, top_k=3):
    """
    Prepare a grounded RAG prompt using context-aware retrieval.
    Does not call Gemini.
    """

    retrieved = search_policies_with_memory(
        query,
        top_k=top_k
    )

    context_parts = []

    for _, row in retrieved.iterrows():
        source = row["source"]

        policy_text = documents_df.loc[
            documents_df["source"] == source, "text"
        ].iloc[0]

        context_parts.append(
            f"Source: {source}\n{policy_text}"
        )

    context = "\n\n---\n\n".join(context_parts)

    history_text = "\n".join(
        f"{message['role'].capitalize()}: {message['content']}"
        for message in conversation_history[-6:]
    )

    prompt = f"""
You are ShopAssist AI, an e-commerce customer support assistant.

Use the conversation history to understand follow-up questions.
Answer policy questions using only the policy information provided.
Do not invent information. If the policies do not answer the question,
clearly state that the information is unavailable.
Cite the relevant policy source when it supports your answer.

CONVERSATION HISTORY:
{history_text if history_text else "No previous conversation."}

RELEVANT POLICIES:
{context}

CURRENT CUSTOMER QUESTION:
{query}

ANSWER:
"""

    return {
        "prompt": prompt,
        "retrieved_documents": retrieved
    }


print("Memory-aware RAG preparation ready.")
```

```python
from pathlib import Path

policy_dir = Path("data/policies")

policies = {

    "delivery_policy.md": """
# Delivery Policy

## Standard Delivery
Standard domestic delivery takes 3–5 working days after the order has been dispatched.

## Express Delivery
Express delivery takes 1–2 working days after dispatch.

## Processing Time
Orders are normally processed within 1–2 working days before dispatch.

## Delivery Delays
Delivery times are estimates and may be affected by severe weather, public holidays, courier disruptions, or other circumstances outside our control.

Customers can contact support if their order has not arrived within the expected delivery timeframe.
""",

    "returns_policy.md": """
# Returns Policy

Customers may return eligible products within 30 days of delivery.

Products must be:
- unused
- undamaged
- in their original packaging
- accompanied by proof of purchase where required

Customers should initiate a return through the customer account or by contacting customer support.

Certain products may be non-returnable, including personalised products, hygiene-sensitive products that have been opened, and final-sale items.

Items that are damaged or incorrect when received are handled under the damaged-items and replacement policies.
""",

    "refund_policy.md": """
# Refund Policy

Refunds are issued after the returned product has been received and inspected.

Once a return is approved, the refund is normally issued to the original payment method.

Customers should allow 5–10 working days for the refund to appear in their account, depending on the payment provider.

Shipping charges are generally non-refundable unless the return is due to an error on our part or the product arrived damaged or incorrect.

If a customer has not received an expected refund, they should contact customer support.
""",

    "replacement_policy.md": """
# Replacement Policy

Customers may request a replacement if they receive:
- the wrong product
- a damaged product
- a defective product
- an item that is missing essential components

Replacement requests should normally be submitted within 30 days of delivery.

If the requested product is unavailable, the customer may be offered a refund instead.

Customers may be asked to provide photographs or other information to verify a damaged or incorrect item.
""",

    "payment_policy.md": """
# Payment Policy

We accept major debit and credit cards and other payment methods displayed during checkout.

Payment must normally be successfully authorised before an order can be processed.

A payment may appear as pending while the payment provider completes authorisation.

If a payment fails, customers should verify their payment details or try another supported payment method.

We do not store complete card numbers on our systems.
""",

    "cancellation_policy.md": """
# Cancellation Policy

Customers may request cancellation before an order has been dispatched.

Once an order has been dispatched, it cannot normally be cancelled.

If an order has already been dispatched, the customer may use the returns process after delivery if the product is eligible for return.

Cancellation requests should be submitted as soon as possible because orders may be processed quickly.
""",

    "account_policy.md": """
# Account Policy

Customers can create an account using a valid email address and password.

Customers can update their name, contact information, password, and other available account details through their account settings.

If a customer forgets their password, they can use the password-reset process.

Customers should contact support if they believe their account has been compromised.

Customers are responsible for keeping their login credentials secure.
""",

    "discount_policy.md": """
# Discount Policy

Promotional discount codes must be entered during checkout before the order is completed.

Only one promotional discount code can normally be applied to an order unless the promotion specifically states otherwise.

Discount codes may have expiry dates, product restrictions, minimum-spend requirements, or other conditions.

Expired or invalid discount codes cannot be applied.

Discounts generally cannot be applied retroactively to completed orders.
""",

    "shipping_policy.md": """
# Shipping Policy

We offer domestic shipping and international shipping to selected countries.

Available shipping methods and their costs are displayed during checkout.

Customers are responsible for providing an accurate delivery address.

Once an order has been dispatched, the customer may receive tracking information when tracking is available.

Shipping times may vary depending on the destination, selected shipping method, customs processing, and courier conditions.
""",

    "warranty_policy.md": """
# Warranty Policy

Eligible products are covered by the manufacturer's warranty where applicable.

Warranty coverage may depend on the product, manufacturer, purchase date, and nature of the issue.

Normal wear and tear, accidental damage, misuse, and unauthorised modifications may not be covered.

Customers should contact support with their order information and a description of the issue to determine whether warranty assistance is available.

Warranty claims may require photographs, product information, or other supporting evidence.
""",

    "gift_cards.md": """
# Gift Card Policy

Gift cards can be used to purchase eligible products through the online store.

Gift cards cannot normally be exchanged for cash.

Gift cards should be treated as valuable payment instruments and kept secure.

If a gift card is lost or stolen, customers should contact support as soon as possible.

Gift card terms, including validity periods and usage restrictions, are displayed at the time of purchase.
""",

    "international_shipping.md": """
# International Shipping Policy

We ship internationally to selected countries.

International shipping costs and available delivery methods are calculated during checkout based on the destination and package.

International deliveries may be subject to customs duties, import taxes, or other charges imposed by the destination country.

Customers are responsible for any applicable customs duties and import charges unless otherwise stated during checkout.

International delivery times may be longer than domestic delivery because of customs processing.
""",

    "damaged_items.md": """
# Damaged Items Policy

Customers should contact support as soon as possible if an item arrives damaged.

Customers may be asked to provide photographs of:
- the damaged product
- the packaging
- the shipping label

A damaged item may qualify for a replacement or refund depending on the circumstances.

Customers should not dispose of the damaged product or packaging until support has advised them to do so.

Damage caused after delivery through misuse or accidental damage may not qualify for a replacement or refund under this policy.
""",

    "missing_items.md": """
# Missing Items Policy

Customers should contact support if an expected item is missing from their delivery.

Before contacting support, customers should check the order confirmation and delivery information to determine whether the order was split into multiple shipments.

If an item was included in the order but is missing from the delivered package, support will investigate the issue.

Depending on the investigation, the missing item may be replaced or refunded.

Customers may be asked to provide photographs of the received package and its contents.
""",

    "privacy_policy.md": """
# Privacy Policy

We collect and process customer information required to provide our services, process orders, provide customer support, and improve our services.

Customer information may include contact details, order information, account information, and information provided during customer support interactions.

Customers should not share passwords, full payment card numbers, or other sensitive authentication information with customer support.

Personal information is handled according to applicable privacy and data-protection requirements.

Customers can contact support with questions about their personal information or privacy rights.
"""
}

# Write all policy files
for filename, content in policies.items():
    file_path = policy_dir / filename
    file_path.write_text(content.strip(), encoding="utf-8")

print(f"Created {len(policies)} policy documents.")
```

```python
# List all policy documents

policy_files = sorted(policy_dir.glob("*.md"))

print(f"Number of policy documents: {len(policy_files)}\n")

for file in policy_files:
    print(file.name)
```

```python
import pandas as pd
from pathlib import Path

policy_dir = Path("data/policies")

# Collect basic information about every policy
policy_data = []

for file_path in sorted(policy_dir.glob("*.md")):
    text = file_path.read_text(encoding="utf-8")

    policy_data.append({
        "filename": file_path.name,
        "characters": len(text),
        "words": len(text.split()),
        "lines": len(text.splitlines())
    })

policy_df = pd.DataFrame(policy_data)

policy_df
```

```python
print("Knowledge Base Statistics")
print("=" * 40)

print(f"Number of documents: {len(policy_df)}")
print(f"Total words: {policy_df['words'].sum():,}")
print(f"Average words per document: {policy_df['words'].mean():.1f}")
print(f"Shortest document: {policy_df['words'].min()} words")
print(f"Longest document: {policy_df['words'].max()} words")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    policy_df["filename"],
    policy_df["words"]
)

plt.xlabel("Policy Document")
plt.ylabel("Number of Words")
plt.title("ShopAssist Knowledge Base — Document Lengths")

plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()
```
